__getitem__ dundermethod

```python
class ShoppingCart:
    def __init__(self,customer_name):
        self.customer_name = customer_name
        self.items = ["Laptop","Mouse","Pendrive"]
    def __len__(self): # Dunder method
        return len(self.items)    
    def __getitem__(self, index):
        return self.items[index]

my_cart = ShoppingCart("nimal")
print(my_cart[1]) 
```

Call dundermethod

```python
class Scalar:
    def __init__(self,factor):
        self.factor = factor
    
    def __call__(self,x):
        return self.factor*x    

my_scalar = Scalar(2)

print(my_scalar(10))        
```

repr Dundermethod

```python
class Person:
    def __init__(self,name,age):
        self.name = name
        self.age = age
    def __repr__(self):
        return f"person_1(name = {self.name} , age = {self.age})"  
      
    def __str__(self):
        return "this is a person class"
person_1 = Person("kamal",23)
print(person_1)
print(repr(person_1))        
```

```python
import math
a = [1,-1,3]
b = [3,2,-2]

def distance_between_vectors(a,b):
    if len(a) == len(b):
        return math.sqrt(sum([math.pow(a[i]-b[i],2) for i in range(len(a))]))
    return("vector length is not match")

print(distance_between_vectors(a,b))    
```

Manhattan distance

```python
a = [-1,3]
b = [-4,-1]

def manhatton_distance(a,b):
    if len(a) == len(b):
        return sum([abs(a[i]-b[i]) for i in range(len(a))])
    return("vector length is not match")
print(manhatton_distance(a,b))    
    
```

```python
a = [-2,3,1]
b = [-3,-2,3]

def get_dotproduct(a,b):
    if len(a) == len(b):
        return sum([a[i]*b[i] for i in range(len(a))])
    return("vector length is not match")
print(get_dotproduct(a,b))    
```
