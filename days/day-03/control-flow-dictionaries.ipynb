```python
scores = [0.91,0.88,None,0.95,0.25,0.72,0.81]
valied_scores = []
proces_count = 0
for i in scores:
    proces_count += 1
    if i == None:
        continue
    elif i < 0.30:
        print("Critical anomaly detected")
        break
    else:
        valied_scores.append(i)
print(valied_scores)
print(proces_count)     
   
```

```python
attempt_results = [0.72,0.81,0.86,0.83,0.94]
threshold = 0.90
max_attempts = 10
atempt = 1

while atempt <= max_attempts and atempt <= len(attempt_results):
        print(f"Attemt {atempt} :accuracy {attempt_results[atempt-1]}")
        
        if attempt_results[atempt-1] >= threshold:
            print("training is succesed")
            break
    
        atempt += 1    
```

Exception handling

```python
raw_values = ["0.83","0.49","bad","0.72"]
clear_scores = []

for value in raw_values:
    try:
        clear_scores.append(float(value))
    except ValueError:
        print(f"value error skiped {value}")
print(clear_scores)        
```
