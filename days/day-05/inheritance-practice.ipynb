Inheritence

```python
class Vehicle:
    def __init__(self,manufacturing_date,color):
        self.manufacuring_date = manufacturing_date
        self.color = color 
        
    has_wheels =True
    
    def display(self):
        print("This is a vehicle")

class Car(Vehicle):
    
    def __init__(self, manufacturing_date, color,brand):
        super().__init__(manufacturing_date, color)
        self.brand = brand
    
    
    def display(self):  # method overriding
        print(f"manufacturing date:{self.manufacuring_date} \ncolor:{self.color} \nbrand:{self.brand}")
        

car = Car("1978/10/20","Red","BMW")
car.display()
               
```

```python
vector = [20,10,30]
def Scalar_multipication(multiply_value):
    value = [vector[i]*multiply_value for i in range(len(vector))] 
    print(value)     
Scalar_multipication(5)        
```

```python
import math
vector = [1,-2,2]

def find_magnitude(vector):
     result = math.sqrt(sum([math.pow(value,2) for value in vector]))  
     print(f" magnitude : {result}")
     
find_magnitude(vector)   
    
```
