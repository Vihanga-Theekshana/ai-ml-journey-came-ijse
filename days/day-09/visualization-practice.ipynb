```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.DataFrame({
    'age':[22,35,28,41,33,26,45,31],
    'score':[0.81,0.44,0.92,0.67,0.70,0.55,0.60,0.88],
    'city':['Colombo','Kandy','Galle','Colombo','Galle','Kandy','Colombo','Galle']
  
})
df
```

```python
plt.figure()
plt.scatter(df['age'],df['score'])
plt.title('Age vs Score')
plt.xlabel('age')
plt.ylabel('score')
plt.show()
```

```python
plt.figure()
sns.boxplot(data=df,x='city',y='score')
plt.title('score spread by city')
plt.show()
```

```python
numeric = df[['age','score']]
corr = numeric.corr()
print(corr)

plt.figure()
sns.heatmap(corr,annot=True,cmap='YlOrBr',vmin=-1,vmax=1)
plt.title('correlation heatmap')
plt.tight_layout()
plt.show()
```
