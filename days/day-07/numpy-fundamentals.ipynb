```python
import numpy as np

array_1 = np.array([9,10,11,12])
array_2 = np.array([6,19,17,13])

def get_dotproduct(array_1,array_2):
    if len(array_1) == len(array_2):
        return sum((array_1) * (array_2))
    print("not match vectors")    

print(get_dotproduct(array_1,array_2))
    
```
