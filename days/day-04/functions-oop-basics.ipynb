Function

```python
def accuracy(correct_lables,total_lables=20):
     accuracy = correct_lables/total_lables
     return accuracy

accuracy_value = accuracy(total_lables=10,correct_lables=8)
print(accuracy_value)
```

```python
import math
def hypotenuse(x_axis,y_axis):
    try:
        if x_axis <= 0 or y_axis <= 0:
            return "Error : value should be positive"
        hypotenuse = math.sqrt(pow(x_axis,2) + pow(y_axis,2))
        return hypotenuse
    except ValueError:
        return "value error"
hypotenuse_value = hypotenuse(4,5)
print(hypotenuse_value)

```

Arbitrary argument

```python
def average(*args):
    if not args:
        return 0
    return sum(args)/len(args)

print(average(1,2,4,6))
print(average(10,20))
```

Keyword aeguments

```python
def average(**keywords):
    print(type(keywords))
    if not keywords:
        return 0.0
    return sum(keywords.values())/len(keywords)
        
        
average_value = average(number_1=3,number_2=6)    
print(average_value)
```

Docstring

```python
import math
def calculate_area(radius):
    """
    Calculate the area of Circle given radius.

    Args:
        radius (float): The radius of the circle.
    Returns:
        float: The calculated area.    
    """
    
    
    
    return math.pi*pow(radius,2)

print(calculate_area.__doc__)
print(calculate_area(5))
```

```python
mylist = [19,20,34,50]

sorted(mylist,key=None,reverse=True)
```

```python
students = [("Jhon",85),("Alice",92),("Bob",76)]
sorted(students,key=lambda value:value[1],reverse=False)
```

```python
users = [
    {"name":"Alice","age":30},
    {"name":"Bob","age":25},
    {"name":"Charlie","age":35},
]

result = sorted(users,key=lambda value:value["age"] , reverse=False)
print(result)
```

```python

minimum = min(users,key=lambda value:value["age"])
print(minimum)
maximum = max(users,key=lambda value:value["age"])
print(maximum)
```

List comprehension

```python
num_list = [1,3,4,6,7]
square_list = [value**2 for value in num_list]
print(square_list)
```

```python
my_list =[8,5,9,4,20]
squared_list = [item**2 for item in my_list if item>5]
print(squared_list)
```

Dictionary comprehension

```python
lables = ['Cat','Dog','Fish','Bear']


mapped_lables = { item:index for index,item in enumerate(lables)}
print(mapped_lables)
```

```python
class Circle:
    def __init__(self,radius):
        self.radius = radius
    
    def calculate_area(self):
        return math.pi*pow(self.radius,2)


circle_1 = Circle(7)
print(circle_1.calculate_area())    

        
```
