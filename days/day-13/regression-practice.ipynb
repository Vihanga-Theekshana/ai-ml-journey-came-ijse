```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

# metrices mean RSS MSE RMSE MAE 
from sklearn import metrics

sns.set_style('whitegrid')
```

```python
df = pd.read_csv("advertising.csv")
print(df.shape)
print(df.head())
print(df.describe())
print(df.info())
print(df.isna().sum())
print(df.duplicated().sum())
```

```python
sns.pairplot(df)
plt.show()
```

```python
x = df[['TV','Radio','Newspaper']]
print(x.shape)
y = df['Sales']
print(y.shape)
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
print(x_train.shape)
print(y_train.shape)
```

```python
simple_model = LinearRegression()
simple_model.fit(x_train,y_train)
```

```python
print(f'Bias : {simple_model.intercept_:.4f}')
print(f'weight_1 : {simple_model.coef_[0]:.4f}')
print(f'weight_2 : {simple_model.coef_[1]:.4f}')
print(f'weight_3 : {simple_model.coef_[2]:.4f}')
```

```python
y_train_pred = simple_model.predict(x_train)
rmse_train = metrics.root_mean_squared_error(y_train,y_train_pred)
print(f'RMSE (Train) :{rmse_train:.4f}')

y_test_pred = simple_model.predict(x_test)
rmse_test = metrics.root_mean_squared_error(y_test,y_test_pred)
print(f'RMSE (Test) :{rmse_test:.4f}')

```

```python
y_train_pred = simple_model.predict(x_train)
mae_train = metrics.mean_absolute_error(y_train,y_train_pred)
print(f'MAE (Train) :{mae_train:.4f}')

y_test_pred = simple_model.predict(x_test)
mae_test = metrics.mean_absolute_error(y_test,y_test_pred)
print(f'MAE (Test) :{mae_test:.4f}')
```

```python
y_train_pred = simple_model.predict(x_train)
r2_train = metrics.r2_score(y_train,y_train_pred)
print(f'R2 (Train) :{r2_train:.4f}')

y_test_pred = simple_model.predict(x_test)
r2_test = metrics.r2_score(y_test,y_test_pred)
print(f'R2 (Test) :{r2_test:.4f}')
```

```python
new_value = pd.DataFrame({'TV':[200.00],
                         'Radio':[30.00],
                         'Newspaper':[10.00]})
result = simple_model.predict(new_value)
print(f'predicted price: {result[0]:.4f}')
```
