Data Visualization

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.DataFrame({
    'age':[22,35,28,41,33,26,45,31],
    'score':[0.81,0.44,0.92,0.67,0.70,0.55,0.60,0.88],
    'city':['Colombo','Kandy','Galle','Colombo','Galle','Kandy','Colombo','Galle']
})

df
```

Histrogram

```python
plt.figure()
plt.hist(df['age'],bins=5,edgecolor='black')
plt.title('Age distribution')
plt.xlabel('age')
plt.ylabel('count')
plt.show()
```

```python
city_mean = df.groupby('city')['score'].mean()
print(city_mean)
```

```python
plt.figure()
city_mean.plot(kind='bar',color='orange',edgecolor='black')
plt.title('Average score by city')
plt.xlabel('city')
plt.ylabel('Average score')
plt.show()
```

```python
plt.figure()
plt.scatter(df['age'],df['score'])
plt.title('Age vs Score')
plt.xlabel('age')
plt.ylabel('score')
plt.show()
```

```python
plt.figure()
sns.boxplot(data=df,x='city',y='score')
plt.title('Age Spreds')
plt.show()
```

Correlation heatmap

```python
numeric = df[['age','score']]
core_metrix = numeric.corr()
print(core_metrix)
 
plt.figure()
sns.heatmap(core_metrix,annot=True,cmap='YlOrBr',vmin=-1,vmax=1)
plt.title('Correlation heatmap')
plt.show()
 
```

Sigmoid function

```python
import numpy as np

def sigmoid(z):
    return 1/ (1 + np.exp(-z))

x=6 
print(sigmoid(x))
```
