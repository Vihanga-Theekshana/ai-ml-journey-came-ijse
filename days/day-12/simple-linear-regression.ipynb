```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

# metrices mean RSS MSE RMSE MAE 
from sklearn import metrics

sns.set_style('whitegrid')


df = pd.read_csv('advertising.csv')
print(df.shape)
print(df.head())
print(df.isna().sum())
print(df.duplicated().sum())
print(df.info())
print(df.describe())

 
```

```python
print("pairplot")
sns.pairplot(df)
plt.show()
```

```python
x = df[['TV']]
print(x.shape)
y = df['Sales']
print(y.shape)
```

split datas for testing

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
#random state mean get random values same for every random_state=42 when i change it it change and when i use again 42 it gives same values fore gived before
print("Testing set shape (x_train) :",x_train.shape)
print("Testing set shape (x_test) :",x_test.shape)
```

```python
simple_model = LinearRegression()
simple_model.fit(x_train,y_train)
```

```python
print(f'Bias (intercept):{simple_model.intercept_:.4f}')
print(f'Coefficient (slope):{simple_model.coef_[0]:.4f}')
```

```python
y_train_pred = simple_model.predict(x_train)
rmse_train = metrics.root_mean_squared_error(y_train,y_train_pred)
print(f'RMSE (Train) :{rmse_train:.4f}')

y_test_pred = simple_model.predict(x_test)
rmse_test = metrics.root_mean_squared_error(y_test,y_test_pred)
print(f'RMSE (Test) :{rmse_test:.4f}')
```

```python
y_train_pred = simple_model.predict(x_train)
mae_train = metrics.mean_absolute_error(y_train,y_train_pred)
print(f'MAE (Train) :{mae_train:.4f}')

y_test_pred = simple_model.predict(x_test)
mae_test = metrics.mean_absolute_error(y_test,y_test_pred)
print(f'MAE (Test) :{mae_test:.4f}')
```

```python
y_train_pred = simple_model.predict(x_train)
r2_train = metrics.r2_score(y_train,y_train_pred)
print(f'R2 (Train) :{r2_train:.4f}')

y_test_pred = simple_model.predict(x_test)
r2_test = metrics.r2_score(y_test,y_test_pred)
print(f'R2 (Test) :{r2_test:.4f}')
```

Inference

```python
new_tv = pd.DataFrame({'TV':[200.00]})
predicted_value = simple_model.predict(new_tv)
print(f'predicted price for $200.00 : {predicted_value[0]:.4f}')
```
