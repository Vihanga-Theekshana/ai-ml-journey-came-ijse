Boolean masking


```python
scores = np.array([0.6,0.9,0.3,0.2,0.75,0.1])

temp_arr = scores > 0.5

print(temp_arr)
print(scores[temp_arr])
```

```python
scores = np.array([0.6,0.9,0.3,0.2,0.75,0.1])
print(scores[scores>0.5])
```

```python
scores = np.array([
    [3,6,4,0.4],
    [1,5,8,0.6],
    [2,9,12,0.8],
    [4,3,7,0.3]
])

print(scores[scores[:,3]>0.5])
```

Linear algebra basics

```python
vector_1 = np.array([3,-1,2])
vector_2 = np.array([2,6,1])

dot_product = vector_1 @ vector_2
print(dot_product)
```

```python
vector_1 = np.array([3,-1,2])
vector_2 = np.array([2,6,1])

print(np.linalg.norm(vector_1))
```

```python
vector_1 = np.array([3,-1,2])
vector_2 = np.array([2,6,1])

cosine_similarity = (vector_1 @ vector_2)/(np.linalg.norm(vector_1)*np.linalg.norm(vector_2))
print(cosine_similarity) 
```

```python
a = np.array([1,5,7])
b = np.array([4,3,10])

c = np.concatenate((a,b),axis=0) #represent axis=1 only 2d array not 1d
print(c)
```
