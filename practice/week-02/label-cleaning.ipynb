```python
raw_lables = [' Positive ','NEGATIVE','neutral','positive',' Negative','pos',' negative ']
clean_lables = []
rejected_lable_count = 0

for lable in raw_lables:
    normalized_lable = lable.strip().lower()
    if 'positive' == normalized_lable :
        clean_lables.append(normalized_lable);
    elif 'negative' == normalized_lable:
        clean_lables.append(normalized_lable);
    else:    
        rejected_lable_count += 1
        continue
print(f'clean_lables = {clean_lables} rejected_count = {rejected_lable_count}')        
```

```python
raw_values = ['0.91','bad',' ','0.72','0.88','0.65','']
clean_values = []

for value in raw_values :
    stripped_value = value.strip()
    if not stripped_value:
        continue
    else:
        try:
            clean_values.append(float(stripped_value))
        except ValueError:
            print(f"message:skipping invalid value : {stripped_value}")
if len(clean_values) == 0:
    print('No valied values')
else:
    print(f"final list = {clean_values} average:{sum(clean_values)/len(clean_values):.3f}")                 
            
```

```python
predictions = ['cat','dog','cat','bird','dog']
true_lables = ['cat','cat','cat','bird','cat']

mistakes_containing = []
mistake_count = 0

for index,value in enumerate(predictions):
    if predictions[index] == true_lables[index]:
        continue
    else:
        mistake_count += 1
        mistakes_containing.append(tuple((index,predictions[index],true_lables[index])))
        
print(f"mistakecount = {mistake_count} mistake with highest index {mistakes_containing[-1]}")  
    
```

Question 11

```python
lables = ['cat','dog','cat','bird','dog','cat','bird','dog','dog','bird']
lable_count = {}
majority_lable = {}
first_index = {}
max_key = ''
max_count = 0

for i,lable in enumerate(lables):
    if lable in lable_count:
        lable_count[lable] += 1
    else:    
        lable_count[lable] = 1
        first_index[lable] = i

for key,value in lable_count.items():
    if value > max_count:
        max_key,max_count = key,value
    elif value == max_count:
        if first_index[key] < first_index[max_key]:
            max_key,max_count = key,value
            

        
majority_lable[max_key] = max_count    
                
print(f" lable_count = {lable_count} \n majority_lable = {majority_lable}")    
```
